## Chapter 1 - Functions

Functions are a tool for describing the real world in mathematical terms. A function can be represented by an equation, a graph, a numerical table, or a verbal description

**Verbal description**

* The interest paid on a cash investment (***y***) depends on the length of time the investment is held (***x***)
* The distance an object travels at constant speed along a straight-line path (***y***) depends on the elapsed time (***x***)

The set of all possible input values is called the **domain** of the function. The set of all output values of the function as x varies throughout the input is called the **range** of the function.

```python
from sympy import symbols, Eq, Piecewise, sin, log, pi
from sympy.plotting import plot, plot3d, plot_implicit
```

```python
# Declare independent variables
x, y = symbols('x y')

# 1. Linear Function (2D)
fx1 = x + 2
plot(fx1, (x, -10, 10), title="f(x) = x + 2")

# 2. Quadratic Function (2D) - This is also called a power function
fx2 = x**2
plot(fx2, (x, -10, 10), title="f(x) = x^2")

# 3. Not a graph of a function - the vertical line intersects the circle at two points
fx3 = Eq(x**2 + y**2, 1)
fx4 = Eq(x,0)
plot_implicit(fx3 |fx4, (x, -1.5, 1.5), (y, -1.5, 1.5), title="Not a function")

# 4. Piece-wise function
## f(x) = x + 2  if x < 0
## f(x) = x^2    if x >= 0
fx5 = Piecewise(
    (x + 2, x < 0),
    (x**2, x >= 0)
)
plot(fx5, (x, -5, 5), title="Piecewise Function")

# 5. Power Function
fx6 = x**(-2)
plot(fx6, (x, -1, 1), title="f(x) = x^-1")

# 6. Rational Function - A rational function is a quotient or ratio ƒ(x) = p(x)/q(x), where p and q are polynomials. The domain of a rational function is the set of all real x for which q(x) != 0.
fx7 = (2*x**2 - 3) / (7*x + 4)
plot(fx7, (x, -4, 4), ylim=(-10, 10), title="Rational Function: f(x) = (2x^2 - 3)/(7x + 4)")

# 7. Trignometric Functions
plot(sin(x), (x, -2*pi, 2*pi), title="f(x) = sin(x)")

# 8. Logarithmic Functions
plot(log(x, 2), (x, 0.1, 10), title="f(x) = log2(x)")
plot(log(x, 10), (x, 0.1, 10), title="f(x) = lg(x)")
plot(log(x), (x, 0.1, 10), title="f(x) = ln(x)")
```

A function ƒ can have only one value ƒ(x) for each x in its domain, so no vertical line can intersect the graph of a function more than once. If a is in the domain of the function ƒ, then the vertical line x = a will intersect the graph of ƒ at the single point (a, ƒ(a)).

Every input in the domain must map to exactly one output in the codomain.

### Monotonicity of Functions

* **Increasing Function:** As $x$ gets bigger, $y$ gets bigger or stays the same ($f(x_1) \le f(x_2)$ when $x_1 < x_2$).
* **Strictly Increasing:** As $x$ gets bigger, $y$ always gets bigger ($f(x_1) < f(x_2)$ when $x_1 < x_2$).
* **Decreasing Function:** As $x$ gets bigger, $y$ gets smaller or stays the same ($f(x_1) \ge f(x_2)$ when $x_1 < x_2$).
* **Strictly Decreasing:** As $x$ gets bigger, $y$ always gets smaller ($f(x_1) > f(x_2)$ when $x_1 < x_2$).

```python
from sympy import symbols, diff, plot, Interval, oo
from sympy.solvers.inequalities import solve_univariate_inequality

x = symbols('x')

# Define your function
f = x**3 - 3*x
print(f"Function: f(x) = {f}")

# Calculate the first derivative
df = diff(f, x)
print(f"First Derivative: f'(x) = {df}\n")

# Find intervals where f(x) is strictly increasing (df > 0)
increasing_intervals = solve_univariate_inequality(df > 0, x, relational=False)
print(f"Strictly Increasing on: {increasing_intervals}")

# Find intervals where f(x) is strictly decreasing (df < 0)
decreasing_intervals = solve_univariate_inequality(df < 0, x, relational=False)
print(f"Strictly Decreasing on: {decreasing_intervals}")

# Plot the function to visualize
plot(f, (x, -3, 3), title="f(x) = x^3 - 3x")
```

```python
from sympy import symbols, simplify

x = symbols('x')
f = x**2

if simplify(f.subs(x, -x) - f) == 0:
  print("Even")
elif simplify(f.subs(x, -x) - f) == f:
  print("Odd")
else:
  print("Neither")
```

Any function constructed from polynomials using algebraic operations (addition, subtraction, multiplication, division, and taking roots) lies within the class of **algebraic functions**. All rational functions are algebraic

**Transcendental functions** are functions that are not algebraic. They include the trigonometric, inverse trigonometric, exponential, and logarithmic functions, and many other functions as well

### Combining Function

```python
from sympy import symbols, plot, sin

x = symbols('x')

# ==============================================================================
# 1. ADDITION
# Example: Total monthly earnings from a fixed base salary plus sales commission.
# Base Function (Salary): f(x) = 2000 (Fixed $2000 base pay)
# Transformation (Addition): Adding g(x) = 0.10*x (10% commission on x dollars of sales)
# ==============================================================================
# Base function: Fixed base pay
f_base_salary = 2000 + 0*x

# Transformed function: Base salary + commission
f_total_earnings = f_base_salary + 0.10 * x


# ==============================================================================
# 2. SUBTRACTION
# Example: Net profit calculated by taking total revenue and subtracting fixed overhead costs.
# Base Function (Revenue): f(x) = 50*x (Selling x items at $50 each)
# Transformation (Subtraction): Subtracting g(x) = 300 (Fixed monthly rent/overhead)
# ==============================================================================
# Base function: Gross revenue
f_revenue = 50 * x

# Transformed function: Net profit (Revenue - Overhead)
f_net_profit = f_revenue - 300


# ==============================================================================
# 3. MULTIPLICATION
# Example: Total revenue made from selling event tickets where price varies with quantity demand.
# Base Function (Price per ticket): f(x) = 100 - 2*x (Price drops as quantity x increases)
# Transformation (Multiplication): Multiplying by g(x) = x (Number of tickets sold)
# ==============================================================================
# Base function: Price demand curve
f_ticket_price = 100 - 2 * x

# Transformed function: Total revenue (Price * Quantity)
f_total_revenue = f_ticket_price * x


# ==============================================================================
# 4. DIVISION
# Example: Average manufacturing cost per item as production scales.
# Base Function (Total Cost): f(x) = 500 + 15*x ($500 setup fee + $15 per unit)
# Transformation (Division): Dividing by g(x) = x (Total units produced)
# ==============================================================================
# Base function: Total production cost
f_total_cost = 500 + 15 * x

# Transformed function: Average cost per unit
f_avg_cost = f_total_cost / x


# ==============================================================================
# 5. VERTICAL SHIFT
# Example: Taxi cab fare structure.
# Base Function (Distance Fare): f(x) = 2.50*x ($2.50 charged per mile x)
# Transformation (Vertical Shift): Adding a $3.50 flat pickup fee: f(x) + 3.50
# ==============================================================================
# Base function: Mileage cost
f_mileage_cost = 2.50 * x

# Transformed function: Total fare (Shifted up by 3.50)
f_taxi_fare = f_mileage_cost + 3.50


# ==============================================================================
# 6. HORIZONTAL SHIFT
# Example: Adjusting a daily temperature model for a different time zone (3 hours later).
# Base Function (Temperature): f(x) = - (x - 12)**2 + 85 (Peaks at hour 12 / noon)
# Transformation (Horizontal Shift): Shift 3 hours right: f(x - 3)
# ==============================================================================
# Base function: Original temperature model
f_temp_local = - (x - 12)**2 + 85

# Transformed function: Adjusted for target time zone (Shifted right by 3)
f_temp_timezone = - ((x - 3) - 12)**2 + 85


# ==============================================================================
# 7. VERTICAL SCALING
# Example: Converting a foreign stock investment portfolio value from USD to EUR.
# Base Function (Stock Portfolio Value in USD): f(x) = 150*x + 50
# Transformation (Vertical Scaling): Multiply by exchange rate 0.92: 0.92 * f(x)
# ==============================================================================
# Base function: Portfolio value in USD
f_portfolio_usd = 150 * x + 50

# Transformed function: Portfolio value in EUR (Vertically scaled by 0.92)
f_portfolio_eur = 0.92 * f_portfolio_usd


# ==============================================================================
# 8. HORIZONTAL SCALING
# Example: Watching a podcast or lecture video at double speed (2x speed).
# Base Function (Audio Signal / Volume over time in minutes): f(x) = sin(x)
# Transformation (Horizontal Scaling): Double the speed compresses time: f(2*x)
# ==============================================================================
# Base function: Normal playback speed
f_audio_normal = sin(x)

# Transformed function: 2x speed playback (Horizontally compressed by factor of 2)
f_audio_fast = sin(2 * x)


# ==============================================================================
# 9. REFLECTION ACROSS X-AXIS
# Example: Converting a profit trajectory into a debt/loss tracking model.
# Base Function (Profit Model): f(x) = 10*x - 20
# Transformation (X-Axis Reflection): Negating output: -f(x)
# ==============================================================================
# Base function: Business profit
f_profit = 10 * x - 20

# Transformed function: Net loss / debt accumulation (Reflected across X-axis)
f_loss = -f_profit


# ==============================================================================
# 10. REFLECTION ACROSS Y-AXIS
# Example: Running an economic projection model backward in time to estimate past trends.
# Base Function (Future Population Growth): f(x) = 1000 * (1.05)**x
# Transformation (Y-Axis Reflection): Negating input domain: f(-x)
# ==============================================================================
# Base function: Future population growth model
f_pop_future = 1000 * (1.05)**x

# Transformed function: Historical population model (Reflected across Y-axis)
f_pop_history = 1000 * (1.05)**(-x)


# ==============================================================================
# PLOTTING EXAMPLES
# ==============================================================================
# Visualizing Taxi Fare (Vertical Shift)
plot(f_mileage_cost, f_taxi_fare, (x, 0, 10),
     title="Vertical Shift: Mileage Cost vs Total Taxi Fare",
     legend=True)

# Visualizing Average Cost per Unit (Division)
plot(f_avg_cost, (x, 1, 50), ylim=(0, 100),
     title="Division: Average Manufacturing Cost per Unit",
     legend=True)
```
