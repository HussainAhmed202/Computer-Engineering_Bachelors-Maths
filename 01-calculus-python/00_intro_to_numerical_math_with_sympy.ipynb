## Chapter 1 - Functions

Functions are a tool for describing the real world in mathematical terms. A function can be represented by an equation, a graph, a numerical table, or a verbal description

**Verbal description**

* The interest paid on a cash investment (***y***) depends on the length of time the investment is held (***x***)
* The distance an object travels at constant speed along a straight-line path (***y***) depends on the elapsed time (***x***)

The set of all possible input values is called the **domain** of the function. The set of all output values of the function as x varies throughout the input is called the **range** of the function.

```python
from sympy import symbols, Eq
from sympy.plotting import plot, plot3d, plot_implicit
```

```python
# Declare independent variables
x, y = symbols('x y')

# 1. Linear Function (2D)
fx1 = x + 2
plot(fx1, (x, -10, 10), title="f(x) = x + 2")

# 2. Quadratic Function (2D)
fx2 = x**2
plot(fx2, (x, -10, 10), title="f(x) = x^2")

# 3. Not a graph of a function - the vertical line intersects the circle at two points
fx3 = Eq(x**2 + y**2, 1)
fx4 = Eq(x,0)
plot_implicit(fx3 |fx4, (x, -1.5, 1.5), (y, -1.5, 1.5), title="Not a function")
```

image.png

A function ƒ can have only one value ƒ(x) for each x in its domain, so no vertical line can intersect the graph of a function more than once. If a is in the domain of the function ƒ, then the vertical line x = a will intersect the graph of ƒ at the single point (a, ƒ(a)).

Every input in the domain must map to exactly one output in the codomain.
