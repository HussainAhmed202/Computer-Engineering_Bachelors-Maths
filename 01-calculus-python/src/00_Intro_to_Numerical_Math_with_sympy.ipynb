## Chapter 1 - Functions

Functions are a tool for describing the real world in mathematical terms. A function can be represented by an equation, a graph, a numerical table, or a verbal description

**Verbal description**

* The interest paid on a cash investment (***y***) depends on the length of time the investment is held (***x***)
* The distance an object travels at constant speed along a straight-line path (***y***) depends on the elapsed time (***x***)

The set of all possible input values is called the **domain** of the function. The set of all output values of the function as x varies throughout the input is called the **range** of the function.

```python
from sympy import symbols
from sympy.plotting import plot
```

```python
x,fx = symbols('x fx')
fx = x + 2
plot(fx)
# for i in range(1,10):
#   print(fx.subs(x,i))
```
